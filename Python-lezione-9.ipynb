## 1. Introduzione alle funzioni

### 1.1 Cos'è una funzione?

Una **funzione** è un blocco di codice autonomo e riutilizzabile che compie un'azione specifica. Una funzione può ricevere dei dati in ingresso (detti parametri), elaborarli attraverso una sequenza di istruzioni e restituire un risultato al programma principale.

In Python abbiamo già utilizzato numerose funzioni integrate (chiamate «built-in»), come `print()`, `len()`, `type()`, `range()`, `sum()`, `min()` e `max()`. In questa lezione impareremo a creare le nostre funzioni personalizzate.

### 1.2 Perché usare le funzioni?

- **Riuso del codice (principio DRY, «Don't Repeat Yourself»)**: evita di scrivere più volte la stessa sequenza di istruzioni. Scrivi il codice una sola volta e richiamalo ovunque sia necessario.
- **Modularità e ordine**: permette di scomporre un problema complesso in sotto-problemi più piccoli, isolati e facili da comprendere.
- **Manutenibilità e collaudo**: se si verifica un errore o se occorre cambiare la logica di un calcolo, è sufficiente intervenire sulla funzione corrispondente senza toccare il resto del programma.
- **Leggibilità**: assegnare un nome chiaro e significativo a una funzione rende il codice autoesplicativo.

### 1.3 Definire ed eseguire una funzione

Per definire una funzione si utilizza la parola chiave `def`, seguita dal **nome della funzione**, da una coppia di parentesi tonde `()` e dal simbolo dei due punti `:`. Il corpo della funzione deve essere sempre **indentato** di quattro spazi.

```python
def nome_della_funzione():
    # Blocco di istruzioni indentato
    istruzione_1
    istruzione_2
```

La semplice definizione crea la funzione ma non esegue le sue istruzioni. Per eseguirla occorre effettuare una **chiamata** (o invocazione) scrivendo il suo nome seguito dalle parentesi tonde.

````python
# Definizione della funzione
def saluta():
    print("Ciao! Benvenuto nella lezione sulle funzioni.")

# Chiamata della funzione
saluta()

# Possiamo chiamarla quante volte vogliamo
saluta()
saluta()
````

### 1.4 Flusso di esecuzione

L'interprete Python legge il file riga per riga dall'alto verso il basso. Quando incontra una riga con `def`, Python registra l'esistenza della funzione e ne memorizza il corpo, ma salta l'esecuzione delle istruzioni interne finché la funzione non viene esplicitamente chiamata.

````python
print("1. Prima della definizione")

def sequenza_di_prova():
    print("3. Dentro la funzione: istruzione A")
    print("4. Dentro la funzione: istruzione B")

print("2. Dopo la definizione, ma prima della chiamata")

# Ora chiamiamo la funzione
sequenza_di_prova()
print("5. Dopo la chiamata della funzione")
````

## 2. Parametri e argomenti

### 2.1 Definire parametri e passare argomenti

Le funzioni diventano strumenti versatili quando possono elaborare dati diversi a ogni esecuzione.

- Un **parametro** è una variabile indicata tra le parentesi tonde nella definizione della funzione (`def`). Funziona come segnaposto per il valore che la funzione si aspetta di ricevere.
- Un **argomento** è il valore concreto fornito alla funzione nel momento in cui viene chiamata.

```python
def saluta_persona(nome):  # 'nome' è il parametro
    print(f"Ciao, {nome}!")

saluta_persona("Chiara")   # "Chiara" è l'argomento
saluta_persona("Matteo")   # "Matteo" è l'argomento
```

````python
def saluta_persona(nome):
    print(f"Ciao, {nome}! Benvenuto al corso.")

saluta_persona("Chiara")
saluta_persona("Matteo")
saluta_persona("Elena")
````

### 2.2 Parametri multipli posizionali

Una funzione può accogliere più parametri separati da virgole. Quando la chiamata avviene senza specificare i nomi dei parametri, gli argomenti vengono associati nell'esatto ordine in cui sono stati elencati nella definizione (argomenti posizionali).

````python
def descrivi_animale(tipo_animale, nome_animale):
    print(f"Ho un {tipo_animale} di nome {nome_animale}.")

# Gli argomenti sono posizionali: il primo va in tipo_animale, il secondo in nome_animale
descrivi_animale("gatto", "Micio")
descrivi_animale("cane", "Luna")

# Attenzione all'ordine: invertire gli argomenti altera il significato
descrivi_animale("Luna", "cane")
````

### 2.4 Parametri con valori di default

Possiamo assegnare un valore predefinito a uno o più parametri. Se chi chiama la funzione omette quell'argomento, Python utilizzerà automaticamente il valore di default specificato.

Regola fondamentale: tutti i parametri con valore di default devono essere posizionati **dopo** i parametri obbligatori (privi di default).

````python
def saluta_con_titolo(nome, saluto="Buongiorno", titolo=""):
    if titolo:
        print(f"{saluto}, {titolo} {nome}!")
    else:
        print(f"{saluto}, {nome}!")

# Usa tutti i valori di default per saluto e titolo
saluta_con_titolo("Rossi")

# Sovrascrive solo il saluto
saluta_con_titolo("Bianchi", saluto="Ciao")

# Sovrascrive entrambi
saluta_con_titolo("Verdi", saluto="Gentilissimo", titolo="Dott.")
````

## 3. Il valore di ritorno: return

### 3.1 Cos'è l'istruzione return?

Fino a questo momento le nostre funzioni si sono limitate a stampare messaggi a schermo tramite `print()`. Nella maggior parte dei casi reali, tuttavia, una funzione deve calcolare un dato e **restituirlo** al programma chiamante, affinché possa essere salvato in una variabile o riutilizzato in altre elaborazioni.

L'istruzione `return` termina immediatamente l'esecuzione della funzione e trasmette il valore specificato a chi ha effettuato la chiamata.

````python
def eleva_al_quadrato(numero):
    return numero ** 2

# Il valore restituito viene salvato in una variabile
risultato = eleva_al_quadrato(6)
print(f"Il quadrato di 6 è: {risultato}")

# Il valore restituito può essere usato direttamente in un'espressione
somma_quadrati = eleva_al_quadrato(3) + eleva_al_quadrato(4)
print(f"Somma dei quadrati (3^2 + 4^2): {somma_quadrati}")
````

### 3.2 Differenza tra print() e return

Confondere `print()` e `return` è uno degli errori più frequenti per chi inizia a programmare in Python.

- `print()` è una funzione che si limita a visualizzare caratteri sul terminale. Non restituisce alcun dato utile al codice: una funzione che termina senza `return` restituisce implicitamente l'oggetto speciale `None`.
- `return` non visualizza nulla sullo schermo, ma restituisce il risultato alla riga di codice che ha invocato la funzione, consentendo di memorizzarlo in una variabile o di utilizzarlo in ulteriori operazioni.

| Caratteristica | `print()` | `return` |
|---|---|---|
| **Scopo principale** | Comunicare con l'utente sullo schermo | Trasmettere un valore al programma chiamante |
| **Effetto su una variabile assegnata** | Assegna `None` | Assegna il valore effettivo calcolato |
| **Effetto sul flusso della funzione** | Continua con l'istruzione successiva | Interrompe subito l'esecuzione della funzione |
| **Riuso nei calcoli successivi** | Impossibile | Possibile e immediato |

````python
# Funzione che stampa solo (senza return)
def somma_errata(a, b):
    print(a + b)

# Funzione corretta con return
def somma_corretta(a, b):
    return a + b

# Proviamo ad assegnare i risultati a una variabile
risultato_errato = somma_errata(10, 20)
print(f"Valore memorizzato da somma_errata: {risultato_errato}")  # Restituisce None!

risultato_corretto = somma_corretta(10, 20)
print(f"Valore memorizzato da somma_corretta: {risultato_corretto}")

# Con return possiamo compiere ulteriori operazioni
totale = risultato_corretto * 2
print(f"Valore raddoppiato: {totale}")
````

### 3.3 Uscita anticipata con return

Poiché l'istruzione `return` interrompe immediatamente l'esecuzione della funzione, possiamo utilizzarla per uscire in anticipo non appena una determinata condizione viene soddisfatta. Questo approccio rende il codice più pulito ed evita annidamenti eccessivi di blocchi condizionali.

````python
def valuta_voto(punteggio):
    # Controllo di validità: uscita anticipata se il valore è scorretto
    if punteggio < 0 or punteggio > 100:
        return "Punteggio non valido: deve essere compreso tra 0 e 100"
    
    if punteggio >= 90:
        return "Eccellente"
    if punteggio >= 70:
        return "Buono"
    if punteggio >= 60:
        return "Sufficiente"
    
    return "Non sufficiente"

print(valuta_voto(95))
print(valuta_voto(65))
print(valuta_voto(120))
````

### 3.4 Restituzione di valori multipli

In Python una funzione può restituire più valori separandoli con una virgola. In realtà Python raggruppa automaticamente questi elementi in una **tupla**, che possiamo poi spacchettare («unpacking») in variabili distinte al momento della chiamata.

````python
def statistiche_base(lista_numeri):
    minimo = min(lista_numeri)
    massimo = max(lista_numeri)
    totale = sum(lista_numeri)
    return minimo, massimo, totale

valori = [14, 5, 29, 3, 42, 18]

# Spacchettamento dei tre valori in variabili distinte
valore_min, valore_max, somma = statistiche_base(valori)

print(f"Minimo: {valore_min}")
print(f"Massimo: {valore_max}")
print(f"Somma complessiva: {somma}")
````

## 4. Scope e visibilità delle variabili

### 4.1 Variabili locali e variabili globali

Lo **scope** (o ambito di visibilità) definisce la parte di programma in cui una variabile è accessibile:

- **Variabile locale**: creata all'interno di una funzione. Nasce quando la funzione viene eseguita e viene distrutta al termine della funzione. È invisibile e inaccessibile dall'esterno.
- **Variabile globale**: creata all'esterno di qualsiasi funzione, nel corpo principale dello script. È accessibile in lettura da qualsiasi punto del file.

````python
messaggio_globale = "Sono accessibile ovunque"

def test_scope():
    messaggio_locale = "Esisto solo dentro test_scope"
    print(messaggio_globale)  # Legge la variabile globale
    print(messaggio_locale)   # Legge la variabile locale

test_scope()

# La riga seguente genererebbe NameError perché messaggio_locale non esiste fuori dalla funzione
# print(messaggio_locale)
````

### 4.2 Mascheramento delle variabili (shadowing)

Se all'interno di una funzione creiamo una variabile locale con lo stesso nome di una variabile globale, Python crea una nuova variabile nello scope locale, lasciando intatta e inalterata quella globale.

````python
valore = 100  # Variabile globale

def modifica_locale():
    valore = 50   # Nuova variabile locale: maschera quella globale
    print(f"Dentro la funzione, valore locale = {valore}")

modifica_locale()
print(f"Nel programma principale, valore globale = {valore}")  # Rimane 100
````

### 4.3 La parola chiave global

Se abbiamo l'esigenza esplicita di modificare una variabile globale dall'interno di una funzione, dobbiamo dichiararla con la parola chiave `global`.

Nota bene: l'uso di `global` va limitato il più possibile. Modificare variabili globali dall'interno di funzioni rende il codice difficile da collaudare e incline a effetti collaterali imprevisti. La prassi consigliata consiste sempre nel passare i valori come parametri e restituire i nuovi dati tramite `return`.

````python
contatore_visite = 0

def registra_accesso():
    global contatore_visite
    contatore_visite += 1

registra_accesso()
registra_accesso()
registra_accesso()
print(f"Accessi totali registrati: {contatore_visite}")  # 3
````

### 4.4 Liste ed elementi mutabili passati a una funzione

Quando passiamo a una funzione un oggetto mutabile (come una lista vista nella Lezione 8), la funzione riceve il riferimento all'oggetto originale. Di conseguenza, le modifiche apportate sul posto (come `.append()`, `.pop()`, o la modifica di un elemento per indice) alterano la lista anche all'esterno della funzione.

````python
def aggiungi_corso(elenco, nuovo_corso):
    elenco.append(nuovo_corso)
    corsi = ["Python", "SQL"]
print(f"Prima: {corsi}")

aggiungi_corso(corsi, "Git")
print(f"Dopo: {corsi}")  # La lista originale è stata modificata!
````

Se desideriamo proteggere la lista originale da modifiche indesiderate, dobbiamo passare una copia esplicita mediante il metodo `.copy()` oppure tramite slicing `[:]`.

````python
def elabora_senza_modificare(elenco):
    copia_elenco = elenco.copy()
    copia_elenco.append("Elemento temporaneo")
    return copia_elenco

dati_originali = [1, 2, 3]
nuovi_dati = elabora_senza_modificare(dati_originali)

print(f"Dati originali intatti: {dati_originali}")
print(f"Nuovi dati elaborati: {nuovi_dati}")
````

## 5. Documentazione e parametri arbitrari

### 5.1 Docstring e la funzione help()

Una **docstring** (stringa di documentazione) è un testo racchiuso tra tre doppi apici `"""..."""` collocato subito sotto la riga di definizione di una funzione. Serve a spiegare a chi legge il codice che cosa fa la funzione, quali parametri accetta e cosa restituisce.

Python include uno strumento integrato per consultare la documentazione: la funzione `help()`.

````python
def calcola_area_cerchio(raggio):
    """Calcola l'area di un cerchio dato il raggio.

    Parametri:
        raggio (float o int): Il raggio del cerchio.

    Restituisce:
        float: L'area calcolata del cerchio.
    """
    pi_greco = 3.1415926535
    return pi_greco * (raggio ** 2)

# Consultiamo la documentazione ufficiale della nostra funzione
help(calcola_area_cerchio)
````

### 5.2 Parametri arbitrari: *args (anteprima)

Talvolta non sappiamo in anticipo quanti argomenti verranno passati a una funzione. Anteponendo un asterisco `*` al nome del parametro (per convenzione `*args`), Python raccoglie tutti gli argomenti posizionali aggiuntivi all'interno di una **tupla**.

````python
def somma_tutti(*numeri):
    totale = 0
    for n in numeri:
        totale += n
    return totale

print(somma_tutti(5, 10))
print(somma_tutti(1, 2, 3, 4, 5))
print(somma_tutti(100, 250, 45, 10))
````

### 5.3 Argomenti arbitrari con nome: **kwargs (anteprima)

Anteponendo due asterischi `**` al nome del parametro (per convenzione `**kwargs`), Python raccoglie tutti gli argomenti nominali aggiuntivi all'interno di un **dizionario** formato da coppie chiave-valore.

````python
def stampa_anagrafica(**dati):
    for chiave, valore in dati.items():
        print(f"{chiave.capitalize()}: {valore}")

stampa_anagrafica(nome="Paolo", citta="Milano", professione="Ingegnere", eta=32)
````
